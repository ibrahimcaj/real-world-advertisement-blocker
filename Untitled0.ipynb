```python
from roboflow import Roboflow

rf = Roboflow(api_key=userdata.get('ROBOFLOW_API_KEY'))
project = rf.workspace(userdata.get('ROBOFLOW_WORKSPACE_ID')).project(userdata.get('ROBOFLOW_PROJECT_ID'))

dataset = project.version(1).download("yolov11")
```

```python
from ultralytics import YOLO
import shutil

# MODEL_OVERRIDE_PATH = None
MODEL_OVERRIDE_PATH = project_directory_path + "/best.pt"

if (MODEL_OVERRIDE_PATH):
  model = YOLO(MODEL_OVERRIDE_PATH)
else:
  model = YOLO("yolo11n.pt")  # n/s/m/l/x
  model.train(data=f"{dataset.location}/data.yaml", epochs=10, imgsz=512)

  # model = YOLO("runs/detect/train-2/weights/best.pt")
  # shutil.copy("runs/detect/train-2/weights/best.pt", project_directory_path + "/best.pt")
```
